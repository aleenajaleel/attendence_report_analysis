# Who are the top 5 most present and least present employees? create a bar chart on both

```python
import pandas as pd
```

```python
df=pd.read_csv("Attendence_report_January (1).csv")
df
```

```python
nn=df[["Present Days","Name"]].sort_values(by='Present Days',ascending=False).loc[0:5]
nn
```

```python
import matplotlib.pyplot as plt
import numpy as np
x=nn['Name']
y=nn['Present Days']
plt.title('barchart')
plt.bar(x,y)
plt.show()
```

```python
kn=df[["Present Days","Name"]].sort_values(by='Present Days',ascending=True).iloc[:5]
kn
```

```python
import matplotlib.pyplot as plt
import numpy as np
x=kn['Name']
y=kn['Present Days']
plt.title('barchart')
plt.bar(x,y)
plt.show()
```

# Heatmap to show correlation

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
attend = pd.read_csv("Attendence_report_January (1).csv")
df1=attend.drop(['Name','Normal Working Hours',' OT Hours ','Late Coming Hours',' Early Going Days '], axis='columns')
corr = df1.corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix of attendence Dataset")
```
