```python
employee_name = input("Enter employee name: ")
employee_ID = int(input("Enter employee ID: "))
basic_salary = float(input("Enter basic salary: "))
hra = float(input("Enter your HRA: "))
travel_allowance = float(input("Enter your travel allowance: "))
pf_deduction = float(input("Enter your PF deduction: "))

total_allowance = hra + travel_allowance
gross_salary = basic_salary + total_allowance
net_salary = gross_salary - pf_deduction

print("Employee name =", employee_name)
print("Employee ID =", employee_ID)
print("Basic salary =", basic_salary)
print("HRA =", hra)
print("Travel allowance =", travel_allowance)
print("PF deduction =", pf_deduction)
print("Total allowance =", total_allowance)
print("Gross salary =", gross_salary)
print("Net salary =", net_salary)
```
